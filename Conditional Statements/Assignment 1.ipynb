```python
num = int(input("enter the number"))

for i in range(1,15):
    result = num*i
    if result % 5 ==0:
        print(result,"multiplication of 5")
    else:
        print(result)
```

```python
import random
# The random module allows us to generate random numbers.
count_six = 0   # to count how many times we get 6

for i in range(10):
    roll = random.randint(1, 6)   # roll the dice
    #Generates a random number between 1 and 6 (like a real dice).
    
    print("Roll", i+1, ":", roll)

    if roll == 6:
        #Checks if the roll is 6.
        count_six += 1  #If yes, increase the counter.

print("Number of times we got  6:", count_six)
```

```python
word = input("Enter a word: ")
def count_vowels_consonants(word):
    vowels = "aeiouAEIOU"  # all vowels (lowercase + uppercase)
    vowel_count = 0
    consonant_count = 0

    for char in word:
        if char.isalpha(): 
         #.isalpha() checks if the character is a letter (ignores numbers, spaces, punctuation).
            if char in vowels:
                vowel_count += 1

            else:
                consonant_count += 1

    return vowel_count, consonant_count

vowels, consonants = count_vowels_consonants(word)
print(f"No of vowels: {vowels}")
print(f"no of consonants: {consonants}")
```

```python
def grade(mark):
    if mark >= 90:
        return "A"
    elif mark >= 80:
        return "B"
    elif mark >= 70:
        return "C"
    elif mark >= 60:
        return "D"
    else:
        return "F"

# input marks as a list
marks = list(map(int, input("Enter marks: ").split()))
#input().split() → splits input by spaces into a list of strings.
#map(int, ...) → converts each string to an integer.
#list(...) → stores the integers as a list.
for m in marks:
    print("Grade:", grade(m))
```
