Train linear regression on a dataset. Show where it fails. Apply polynomial
regression and compare results using plots. Write a short explanation

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score

# Step 1: Create Non-Linear Dataset


np.random.seed(0)

X = np.linspace(-3, 3, 100).reshape(-1, 1)
y = 2*X**2 + 3*X + 5 + np.random.randn(100, 1)*3   

# Step 2: Linear Regression

linear_model = LinearRegression()
linear_model.fit(X, y)
y_pred_linear = linear_model.predict(X)

# Step 3: Polynomial Regression (Degree 2)

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

poly_model = LinearRegression()
poly_model.fit(X_poly, y)
y_pred_poly = poly_model.predict(X_poly)

# Step 4: Calculate R² Scores

r2_linear = r2_score(y, y_pred_linear)
r2_poly = r2_score(y, y_pred_poly)

# Step 5: Plot Results


plt.figure()
plt.scatter(X, y)
plt.plot(X, y_pred_linear)
plt.title("Linear Regression")
plt.show()

plt.figure()
plt.scatter(X, y)
plt.plot(X, y_pred_poly)
plt.title("Polynomial Regression (Degree 2)")
plt.show()

print("Linear Regression R2 Score:", r2_linear)
print("Polynomial Regression R2 Score:", r2_poly)
```

Linear regression fails when the relationship between variables is nonlinear because it can only fit a straight line. In such cases, polynomial regression is used to capture curved patterns by adding higher-degree features like x². The plot comparison shows that polynomial regression fits the data better and produces more accurate predictions.
